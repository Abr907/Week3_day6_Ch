```python
#Step 2 Implement word_frequency Method
class Text:
    def __init__(self, text):
        self.text = text

    def word_frequency(self, word):
        # Split the text into a list of words
        words = self.text.split()

        # Count occurrences of the given word (case-insensitive)
        count = sum(1 for w in words if w.lower() == word.lower())

        # Return the count or a message if the word is not found
        if count == 0:
            return None
        return count


# Example usage
text = Text("Python is great. Python is powerful. I love Python.")

print(text.word_frequency("Python"))  # Output: 3
print(text.word_frequency("is"))      # Output: 2
print(text.word_frequency("Java"))    # Output: None
```

```python
#Step 3: Implement most_common_word Method
class Text:
    def __init__(self, text):
        self.text = text

    def most_common_word(self):
        words = self.text.lower().split()

        word_frequency = {}

        for word in words:
            word_frequency[word] = word_frequency.get(word, 0) + 1

        if not word_frequency:
            return None

        most_common = max(word_frequency, key=word_frequency.get)

        return most_common


# Example
text = Text("Prisca and Apolo")
print(text.most_common_word())
```

```python
#Step 4: Implement the unique_words() Method
class Text:
    def __init__(self, text):
        self.text = text

    def unique_words(self):
        words = self.text.split()
        unique = set(words)
        return list(unique)


# Example
text = Text("python is easy and python is powerful")

print(text.unique_words())
```

```python

class Text:
    def __init__(self, text):
        self.text = text

    def word_frequency(self, word):
        words = self.text.split()
        count = words.count(word)
        return count if count > 0 else None

    def most_common_word(self):
        words = self.text.split()

        if not words:
            return None

        frequency = {}

        for word in words:
            frequency[word] = frequency.get(word, 0) + 1

        return max(frequency, key=frequency.get)

    def unique_words(self):
        return list(set(self.text.split()))

    @classmethod
    def from_file(cls, file_path):
        with open(file_path, "r", encoding="utf-8") as file:
            content = file.read()
        return cls(content)


class TextModification(Text):
    pass


# Test the inheritance
text = TextModification("Python is great and Python is powerful")

print(text.word_frequency("Python"))  # 2
print(text.most_common_word())        # Python
print(text.unique_words())            # Unique words
```

```python
# Step 7: Implement remove_punctuation Method
import string

class Text:
    def __init__(self, text):
        self.text = text

    def remove_punctuation(self):
        punctuation = string.punctuation
        modified_text = self.text.translate(
            str.maketrans("", "", punctuation)
        )
        return modified_text

```

```python
text = Text("Hello, world! How are you today?")
print(text.remove_punctuation())
```

```python
# Step 8: Implement remove_stop_words Method

# Define TextModification inheriting from Text and implementing the new method
class TextModification(Text):
    def remove_stop_words(self):
        # Common English stop words
        stop_words = {
            "a", "an", "the", "is", "are", "was", "were",
            "am", "be", "been", "being", "and", "or", "but",
            "if", "in", "on", "at", "to", "of", "for",
            "with", "by", "from", "as", "it", "this", "that",
            "these", "those", "i", "you", "he", "she", "we",
            "they", "my", "your", "his", "her", "our", "their",
            "do", "does", "did", "have", "has", "had", "not"
        }

        # Split the text into words
        words = self.text.split()

        # Remove stop words (case-insensitive)
        filtered_words = [
            word for word in words
            if word.lower() not in stop_words
        ]

        # Join the remaining words into a string
        return " ".join(filtered_words)

text = "This is Barnabas Apolo."
modified_text = TextModification(text)
print(modified_text.remove_stop_words())
```

```python
# Step 9: Implement remove_special_characters() Method
import re

class TextModification(Text):

    def remove_special_characters(self):
        modified_text = re.sub(r'[^a-zA-Z0-9\s]', '', self.text)
        return modified_text
```
